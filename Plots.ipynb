### Molecule image

```python
def generate_mol_images(natoms=14, nfiles=10, save=False):
    i_train, i_test, i_val, _, y_test, _, mean,R_train,R_test,X_train,X_test = ANI_index_returns()
    Z = np.load('/home/vvinod/2025/BigDatasets/ANI1x/ANI1x_multifidelity_50k_allfids.npz',allow_pickle=True)
    Z = Z['Z']
    Ztest = Z[i_test]
    atom_counts = np.array([len(molecule) for molecule in Ztest])
    y_preds = np.load('outputs/final_predictions/ANI_cascading_predictions.npy')
    y_preds_sf = np.load('outputs/final_predictions/ANI_SF_predictions.npy')
    y_preds_ani = np.load('outputs/custom_ani_preds.npy')
    # y_preds_allani = np.load('outputs/pretrained_ani_preds.npy')
    error = np.abs(y_test[:,-1]-y_preds)
    error_sf = np.abs(y_test[:,-1]-y_preds_sf)
    yy = np.load('/home/vvinod/2025/BigDatasets/ANI1x/test_set.npz',allow_pickle=True)['E']
    error_ani = np.abs(yy-y_preds_ani)
    selected_arr = np.where(atom_counts==natoms)[0]
    
    print('MAE (Adaptive): ', np.mean(error),
          '\nMAE (SF): ', np.mean(error_sf),
          '\nMAE (ANI): ', np.mean(error_ani),
          '\nTotal Mols: ',selected_arr.shape)
    if nfiles>1:
        screened_selected_arr = selected_arr[::selected_arr.shape[0]//(nfiles-1)]
    else:
        screened_selected_arr = selected_arr
    try:
        print('\nSelectedMols: ', screened_selected_arr.shape)
    except:
        pass
    if save:
        for i in screened_selected_arr:
            _ = r_z_to_2d_image(R_test[i], Ztest[i], 
                                  mae_mfml=error[i],
                                  mae_sfml=error_sf[i], 
                                  mae_ANI=error_ani[i], 
                                  save_path=f'plots/molimages/{Ztest[i].shape[0]}_{i}_mol.png')
```

## Training Set Size Evolution

```python
def plot_ntrain(mol='CH3Cl',method='satMFML',ax=None,leg=False):
    ax = ax or plt.gca()
    aaa = np.load(f'outputs/{mol}_{method}_trainsizes.npy')
    bbb = np.load(f'outputs/{mol}_{method}_mae.npy')
    # print(aaa.shape)
    # start=1
    colors=plt.cm.viridis(np.linspace(0,1,4))#['b','orange','g','r']
    markers=['+','.','1','x']
    r = np.arange(1,1+aaa.shape[0])
    for i in range(4):
        ax.loglog(r,aaa[:,i],color=colors[i],label=f'{i+1}')#,marker=markers[i])
        # ax.loglog(aaa[:,i],bbb,color=colors[i],marker=markers[i])
        # start += aaa.shape[0]
    # ax.set_xlabel('Epoch')
    ax.set_xlabel('Epoch')
    ax.set_ylabel('$N_{\mathrm{train}}^{f}$')
    if leg:
        ax.legend(['0','1','2','3'])
```

```python
def n_MFML_plot(ax=None):
    ax = ax or plt.gca()
    colors=plt.cm.viridis(np.linspace(0,1,4))#['b','orange','g','r']
    markers=['+','.','1','x']
    ####
    n_trains = []
    for i in range(1,10):
        arr = np.asarray([8,4,2,1])*2**i
        n_trains.append(arr)
        # print(arr)
    n_trains = np.asarray(n_trains)
    ####
    a = np.arange(1,10)
    for i in range(4):
        ax.loglog(a,n_trains[:,i],color=colors[i],label=f'{i+1}',marker='o')
    ax.set_xlabel('Epoch')
    ax.set_ylabel('$N_{\mathrm{train}}^{f}$')
# n_MFML_plot()
```

```python
fig,ax = plt.subplots(1,3,sharey=True,sharex=False,figsize=(12,3))

n_MFML_plot(ax=ax[0])
plot_ntrain('CH3F','globalpass',ax=ax[1])
plot_ntrain('CH3F','cascading',ax=ax[2])
ax[0].set_title('MFML')
ax[1].set_title('Adaptive-MFML')
ax[2].set_title('Ceiling-Adaptive-MFML')
ax[2].legend(title='Fidelity')
# ax[1].set_xlim(100,200)
plt.savefig('plots/trainsizeevolution.pdf',format='pdf',bbox_inches='tight')
```

## Validation Set Vs Test Set MAEs

```python
def plot_val_test_lc_comparison(molname,ax=None):
    ax = ax or plt.gca()
    a = np.load(f'outputs/{molname}_globalpass_mae.npy')
    b = np.load(f'outputs/{molname}_globalpass_valmae.npy')
    r = np.arange(1,1+a.shape[0])
    ax.loglog(r,a,label='Test MAE',linestyle=':')
    ax.loglog(r,b,label='Validation MAE',linestyle='--')
    # ax.legend()
    ax.set_title(molname)
    ax.set_xlabel('Epoch')
    ax.set_ylabel('MAE [kcal/mol]')
    # ax[0].set_yscale('log')
```

```python
fig = plt.figure(figsize=(10, 5))
gs = fig.add_gridspec(2, 3)
ax1 = fig.add_subplot(gs[0, 0])
ax2 = fig.add_subplot(gs[0, 1])
ax3 = fig.add_subplot(gs[1, 0])
ax4 = fig.add_subplot(gs[1, 1])
ax5 = fig.add_subplot(gs[0, 2]) 
plot_val_test_lc_comparison('CH3Cl',ax=ax1)
plot_val_test_lc_comparison('CH3F',ax=ax2)
plot_val_test_lc_comparison('SCF',ax=ax3)
plot_val_test_lc_comparison('EV',ax=ax4)
plot_val_test_lc_comparison('ANI',ax=ax5)
plt.tight_layout(pad=1)
ax4.legend()
plt.savefig('plots/val_test_maecomparison.pdf',format='pdf',bbox_inches='tight')
```
